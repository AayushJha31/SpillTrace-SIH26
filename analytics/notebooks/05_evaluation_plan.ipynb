# SpillTrace Evaluation Plan

## Objective

Evaluate the SpillTrace system at two development milestones:

- Day 10: Early prototype validation
- Day 30: Full system evaluation

The evaluation focuses on segmentation quality, spill geometry,
AIS integration, and system-level performance.

## Day 10 — Early Prototype Evaluation

### 1. Segmentation Quality

Measure:

- IoU
- Dice Score
- Precision
- Recall

Purpose:
Determine whether the initial oil-spill segmentation is detecting
the spill region reasonably well.

### 2. Spill Geometry

Measure:

- Slick area
- Centroid
- Perimeter

Purpose:
Verify that the detected spill geometry can be calculated correctly
from the segmentation mask.

### 3. AIS Data Quality

Measure:

- Percentage of AIS track gaps
- Maximum AIS gap duration

Purpose:
Determine whether the available AIS data is sufficiently complete
for vessel-track analysis.

### Day 10 Success Criteria

- Segmentation metrics can be calculated reliably.
- Geometry calculations pass validation against known values.
- AIS completeness metrics can be generated.
- The complete evaluation pipeline runs on sample data.

## Day 30 — Full System Evaluation

### 1. Segmentation Performance

Measure:

- IoU
- Dice Score
- Precision
- Recall

Compare performance across the available validation samples.

### 2. Spill Geometry Accuracy

Measure:

- Slick area
- Centroid
- Perimeter

Compare predicted spill geometry against available reference
masks/known values where applicable.

### 3. AIS Integration

Measure:

- Percentage of AIS track gaps
- Maximum gap duration
- Quality/completeness of vessel tracks used for analysis

### 4. End-to-End Evaluation

Evaluate whether SAR-based spill detection and AIS vessel-track
information can be combined consistently for the demo scenario.

### Day 30 Success Criteria

- Segmentation performance is quantitatively evaluated.
- Spill geometry is calculated consistently.
- AIS track completeness is quantified.
- SAR and AIS outputs can be evaluated together.
- Results are reproducible using the project's evaluation pipeline.

## Day 10 vs Day 30

| Evaluation Area | Day 10 | Day 30 |
|---|---|---|
| Segmentation | Initial IoU, Dice, Precision, Recall | Full validation and comparison |
| Geometry | Area, centroid, perimeter validation | Geometry evaluation on validation samples |
| AIS | Gap percentage, maximum gap | Track quality and integration |
| Integration | Pipeline runs on sample data | End-to-end SAR + AIS evaluation |
| Goal | Validate prototype | Evaluate complete system |
