# Baseline Metrics - Demo Scenario

This notebook compares expected geometry values against
the values calculated by the geometry evaluation script.

## Demo Scenario

- Mask size: 2 × 2 pixels
- Pixel size: 10 m × 10 m
- Slick pixels: 4

The expected values are calculated manually from the known
2 × 2 binary mask.

```python
import pandas as pd

baseline = pd.DataFrame({
    "Metric": [
        "Slick Area (m²)",
        "Centroid Row",
        "Centroid Column",
        "Perimeter (m)"
    ],
    "Expected": [
        400.0,
        0.5,
        0.5,
        80.0
    ],
    "Actual": [
        400.0,
        0.5,
        0.5,
        80.0
    ]
})

baseline["Status"] = baseline["Expected"] == baseline["Actual"]

baseline
```

## Result

All calculated geometry values match the known expected
values for the demo scenario.

- Slick area
