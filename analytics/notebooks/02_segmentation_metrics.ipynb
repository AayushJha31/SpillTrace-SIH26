# Segmentation Evaluation Metrics

This notebook defines the evaluation metrics used for oil-spill segmentation:

- Intersection over Union (IoU)
- Dice Score
- Precision
- Recall

```python
def segmentation_metrics(y_true, y_pred):
    """
    Calculate IoU, Dice, Precision, and Recall
    for binary segmentation masks.
    """

    y_true = y_true.astype(bool)
    y_pred = y_pred.astype(bool)

    tp = (y_true & y_pred).sum()
    fp = (~y_true & y_pred).sum()
    fn = (y_true & ~y_pred).sum()

    iou = tp / (tp + fp + fn) if (tp + fp + fn) > 0 else 0
    dice = (2 * tp) / (2 * tp + fp + fn) if (2 * tp + fp + fn) > 0 else 0
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0

    return {
        "IoU": iou,
        "Dice": dice,
        "Precision": precision,
        "Recall": recall
    }
```

```python
import numpy as np

y_true = np.array([
    [1, 1, 0],
    [1, 0, 0],
    [0, 0, 1]
])

y_pred = np.array([
    [1, 1, 0],
    [1, 1, 0],
    [0, 0, 0]
])

metrics = segmentation_metrics(y_true, y_pred)

print(metrics)
```
